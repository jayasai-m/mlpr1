```python
# ============================================================
# RESTAURANT REVENUE PREDICTION - COMPLETE GOOGLE COLAB CODE
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

from sklearn.model_selection import train_test_split, KFold, cross_val_score
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LinearRegression, Ridge
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

# ============================================================
# 1. LOAD DATASET FROM GITHUB
# ============================================================

url = "https://raw.githubusercontent.com/allenkong221/restaurant-revenue-prediction/master/train.csv"

try:
    df = pd.read_csv(url)
except:
    url = "https://raw.githubusercontent.com/yc1909/Restaurant-Revenue-Prediction/main/notebook/data/Restaurant_revenue.csv"
    df = pd.read_csv(url)

print("=" * 70)
print("RESTAURANT REVENUE PREDICTION")
print("=" * 70)

print("\nDataset loaded successfully!")
print("Dataset shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

display(df.head())

# ============================================================
# 2. AUTOMATICALLY FIND REVENUE COLUMN
# ============================================================

revenue_column = None

for col in df.columns:
    if col.strip().lower() == "revenue":
        revenue_column = col
        break

if revenue_column is None:
    raise ValueError(
        "Revenue column was not found. Available columns are:\n"
        + str(df.columns.tolist())
    )

print("\nTarget column detected:", revenue_column)

# ============================================================
# 3. BASIC DATA INFORMATION
# ============================================================

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nStatistical Summary:")
display(df.describe(include="all").T)

# ============================================================
# 4. DATA CLEANING
# ============================================================

# Remove ID column
if "Id" in df.columns:
    df = df.drop("Id", axis=1)

# Convert Open Date into useful features
if "Open Date" in df.columns:

    df["Open Date"] = pd.to_datetime(
        df["Open Date"],
        format="%m/%d/%Y",
        errors="coerce"
    )

    df["Open_Year"] = df["Open Date"].dt.year
    df["Open_Month"] = df["Open Date"].dt.month

    # Approximate restaurant age
    df["Restaurant_Age"] = 2015 - df["Open_Year"]

    df = df.drop("Open Date", axis=1)

# ============================================================
# 5. EXPLORATORY DATA ANALYSIS
# ============================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    df[revenue_column],
    kde=True
)

plt.title("Restaurant Revenue Distribution")
plt.xlabel("Revenue")
plt.ylabel("Number of Restaurants")
plt.show()

# Boxplot
plt.figure(figsize=(10, 5))

sns.boxplot(
    x=df[revenue_column]
)

plt.title("Restaurant Revenue Boxplot")
plt.xlabel("Revenue")
plt.show()

# ============================================================
# 6. CORRELATION MATRIX
# ============================================================

numeric_df = df.select_dtypes(
    include=np.number
)

plt.figure(figsize=(14, 10))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")
plt.show()

# ============================================================
# 7. FEATURES AND TARGET
# ============================================================

X = df.drop(
    revenue_column,
    axis=1
)

y = df[revenue_column]

print("\nFeatures:")
print(X.columns.tolist())

print("\nTarget:")
print(revenue_column)

# ============================================================
# 8. IDENTIFY NUMERICAL AND CATEGORICAL FEATURES
# ============================================================

numeric_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("\nNumerical Features:")
print(numeric_features)

print("\nCategorical Features:")
print(categorical_features)

# ============================================================
# 9. TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))

# ============================================================
# 10. PREPROCESSING
# ============================================================

preprocessor = ColumnTransformer(
    transformers=[
        (
            "numerical",
            StandardScaler(),
            numeric_features
        ),
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        )
    ]
)

# ============================================================
# 11. REGRESSION MODELS
# ============================================================

models = {

    "Linear Regression":
        LinearRegression(),

    "Ridge Regression":
        Ridge(alpha=10),

    "Random Forest Regression":
        RandomForestRegressor(
            n_estimators=300,
            max_depth=8,
            random_state=42
        ),

    "Gradient Boosting Regression":
        GradientBoostingRegressor(
            n_estimators=200,
            learning_rate=0.05,
            max_depth=3,
            random_state=42
        )
}

# ============================================================
# 12. TRAIN AND TEST MODELS
# ============================================================

results = {}
trained_models = {}
all_predictions = {}

for name, model in models.items():

    pipeline = Pipeline(
        steps=[
            ("preprocessing", preprocessor),
            ("model", model)
        ]
    )

    pipeline.fit(
        X_train,
        y_train
    )

    y_pred = pipeline.predict(
        X_test
    )

    mae = mean_absolute_error(
        y_test,
        y_pred
    )

    mse = mean_squared_error(
        y_test,
        y_pred
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        y_test,
        y_pred
    )

    results[name] = {
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2 Score": r2
    }

    trained_models[name] = pipeline
    all_predictions[name] = y_pred

# ============================================================
# 13. MODEL COMPARISON
# ============================================================

results_df = pd.DataFrame(results).T

results_df = results_df.sort_values(
    by="RMSE",
    ascending=True
)

print("\n" + "=" * 70)
print("REGRESSION MODEL COMPARISON")
print("=" * 70)

display(
    results_df.style.format({
        "MAE": "{:,.2f}",
        "MSE": "{:,.2f}",
        "RMSE": "{:,.2f}",
        "R2 Score": "{:.4f}"
    })
)

# ============================================================
# 14. 5-FOLD CROSS VALIDATION
# ============================================================

print("\n" + "=" * 70)
print("5-FOLD CROSS VALIDATION")
print("=" * 70)

kf = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_results = []

for name, model in models.items():

    pipeline = Pipeline(
        steps=[
            ("preprocessing", preprocessor),
            ("model", model)
        ]
    )

    scores = cross_val_score(
        pipeline,
        X,
        y,
        cv=kf,
        scoring="neg_root_mean_squared_error"
    )

    rmse_scores = -scores

    cv_results.append({
        "Model": name,
        "Mean CV RMSE": rmse_scores.mean(),
        "Std CV RMSE": rmse_scores.std()
    })

cv_df = pd.DataFrame(cv_results)

cv_df = cv_df.sort_values(
    by="Mean CV RMSE"
)

display(
    cv_df.style.format({
        "Mean CV RMSE": "{:,.2f}",
        "Std CV RMSE": "{:,.2f}"
    })
)

# ============================================================
# 15. FIND BEST MODEL
# ============================================================

best_model_name = results_df.index[0]

best_model = trained_models[
    best_model_name
]

best_prediction = all_predictions[
    best_model_name
]

best_result = results_df.iloc[0]

print("\n" + "=" * 70)
print("BEST MODEL")
print("=" * 70)

print("Best Model :", best_model_name)
print("MAE       :", round(best_result["MAE"], 2))
print("MSE       :", round(best_result["MSE"], 2))
print("RMSE      :", round(best_result["RMSE"], 2))
print("R² Score  :", round(best_result["R2 Score"], 4))

# ============================================================
# 16. RMSE COMPARISON
# ============================================================

plt.figure(figsize=(10, 6))

sns.barplot(
    data=results_df.reset_index(),
    x="RMSE",
    y="index"
)

plt.title("RMSE Comparison of Regression Models")
plt.xlabel("RMSE - Lower is Better")
plt.ylabel("Model")
plt.show()

# ============================================================
# 17. R² COMPARISON
# ============================================================

plt.figure(figsize=(10, 6))

sns.barplot(
    data=results_df.reset_index(),
    x="R2 Score",
    y="index"
)

plt.title("R² Score Comparison of Regression Models")
plt.xlabel("R² Score - Higher is Better")
plt.ylabel("Model")
plt.show()

# ============================================================
# 18. CROSS-VALIDATION COMPARISON
# ============================================================

plt.figure(figsize=(10, 6))

sns.barplot(
    data=cv_df,
    x="Mean CV RMSE",
    y="Model"
)

plt.title("5-Fold Cross Validation RMSE")
plt.xlabel("Mean CV RMSE - Lower is Better")
plt.ylabel("Model")
plt.show()

# ============================================================
# 19. ACTUAL VS PREDICTED
# ============================================================

plt.figure(figsize=(9, 7))

plt.scatter(
    y_test,
    best_prediction,
    alpha=0.7
)

minimum = min(
    y_test.min(),
    best_prediction.min()
)

maximum = max(
    y_test.max(),
    best_prediction.max()
)

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    linestyle="--"
)

plt.title(
    "Actual vs Predicted Revenue\n"
    + best_model_name
)

plt.xlabel("Actual Revenue")
plt.ylabel("Predicted Revenue")

plt.show()

# ============================================================
# 20. RESIDUAL ANALYSIS
# ============================================================

residuals = y_test - best_prediction

plt.figure(figsize=(10, 6))

sns.histplot(
    residuals,
    kde=True
)

plt.title(
    "Residual Distribution - "
    + best_model_name
)

plt.xlabel("Residual")
plt.ylabel("Frequency")

plt.show()

# ============================================================
# 21. ACTUAL VS PREDICTED TABLE
# ============================================================

prediction_table = pd.DataFrame({
    "Actual Revenue": y_test.values,
    "Predicted Revenue": best_prediction
})

prediction_table["Error"] = (
    prediction_table["Actual Revenue"]
    - prediction_table["Predicted Revenue"]
)

print("\nActual vs Predicted Revenue:")

display(
    prediction_table.head(15)
)

# ============================================================
# 22. FINAL REPORT
# ============================================================

print("\n" + "=" * 70)
print("FINAL PROJECT RESULT")
print("=" * 70)

print("""
Objective:
Predict restaurant revenue using machine learning regression models.

Models Tested:
1. Linear Regression
2. Ridge Regression
3. Random Forest Regression
4. Gradient Boosting Regression

Evaluation Metrics:
• MAE
• MSE
• RMSE
• R² Score
• 5-Fold Cross Validation

Best Model:
""")

print(">>>", best_model_name)

print(
    "\nBest RMSE:",
    round(best_result["RMSE"], 2)
)

print(
    "Best R² Score:",
    round(best_result["R2 Score"], 4)
)

print("\nConclusion:")
print(
    "The model with the lowest RMSE on the test dataset "
    "was selected as the best-performing regression model."
)

print("=" * 70)
```
